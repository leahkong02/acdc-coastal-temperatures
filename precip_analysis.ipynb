```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load('P_freq_001_03.npy')
precip_freq = np.linspace(0.01, 0.3, 10) # in days^-1
precip_timescale = [1/x for x in precip_freq]
```

```python
data.shape
```

```python
fig, ax = plt.subplots(1, 1)
plt.boxplot(data.T, tick_labels = np.round(precip_freq, 2), manage_ticks=True)
plt.xticks(rotation=45)
plt.xlabel('a1 - precip freq [days^-1]')
plt.ylabel('skewness')
```

```python
fig, ax = plt.subplots(1, 1)
plt.boxplot(data.T, tick_labels = np.round(precip_timescale, 2), manage_ticks=True)
plt.xticks(rotation=45)
plt.xlabel('Days between rain')
plt.ylabel('')
```
